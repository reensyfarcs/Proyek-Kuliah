DATA VISUALIZATION OF IRIS FLOWER

```python
#Library for data manipulation
import numpy as np
import pandas as pd

#Library for data visualizatin
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = sns.load_dataset("iris")
dataset.info()
```

```python
print(dataset)
```

```python
dataset["species"].unique()
```

```python
fig, ax = plt.subplots(figsize=(8,4))
sns.countplot(data=dataset, x="species", hue="species")

ax.set_title("Barplot iris for count species")
ax.set_ylabel("")
ax.set_xlabel("")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,4))
sns.scatterplot(data=dataset, x="petal_length", y="petal_width", hue="species")

ax.set_title("Barplot iris for count species")
ax.set_ylabel("")
ax.set_xlabel("")

plt.show()
```

```python
#Create heatmap corr
fig, ax = plt.subplots(figsize=(8,4))
sns.heatmap(
    dataset.corr(numeric_only=True), vmin=-1, vmax=1,
    cmap="viridis", annot=True, fmt=".3f", linewidths=1
)

#Set labels
ax.set_title("", fontsize=14)
ax.set_xlabel("", fontsize=12)
ax.set_ylabel("", fontsize=12)
ax.grid(False)

#Show plot
plt.tight_layout()
plt.show()
```
