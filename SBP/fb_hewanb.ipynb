```python
rules = [
    ({"memiliki bulu", "melahirkan"}, "Mamalia"),
    ({"bertelur", "memiliki paruh"}, "Burung"),
    ({"Mamalia", "memakan daging"}, "Karnivora"),
    ({"Karnivora", "mengerang"}, "Singa"),
    ({"Mamalia", "bertaring"}, "Serigala"),
]
```

```python
facts = {
    "memiliki bulu",
    "melahirkan",
    "memakan daging",
    "mengerang"
}
```

```python
def backward_chaining(goal, facts, rules, visited=None):
    if visited is None :
        visited = set()

    if goal in facts :
        print(f"[V] {goal} ditemukan dalam fakta awal")
        return True

    if goal in visited :
        return False
    visited.add(goal)

    for conditions, conclusion in rules :
        if conclusion == goal :
            print (f"[?] Mencoba membuktikan {goal} dengan aturan : {conditions} -> {conclusion}")

            if all (backward_chaining(cond, facts, rules, visited) for cond in conditions):
                print(f"[+] Berhasil membuktikan {goal}")
                facts.add(goal)
                return True
            
    print(f"[x] Tidak dapat membuktikan {goal}")
    return False
```

```python
goal = "Singa"
print("Fakta awal : ", facts)
print("\n--- Proses Backward Chaining ---")
result = backward_chaining(goal, facts, rules)
```

```python
print("\nHasil:", result)
print("Fakta akhir:", facts)
```
