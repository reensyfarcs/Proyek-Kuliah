```python
rules = [
    {
    'id': 'R1',
    'conditions': ['langit_mendung', 'kelembaban_tinggi'],
    'result': 'cuaca_hujan'
    },
    {
    'id': 'R2',
    'conditions': ['cuaca_hujan', 'suhu_rendah'],
    'result': 'bawa_payung'    
    },
    {
    'id': 'R3',
    'conditions': ['langit_cerah', 'suhu_tinggi'],
    'result': 'cuaca_panas'
    },
    {
    'id': 'R4',
    'conditions': ['cuaca_panas'],
    'result': 'pakai_topi'
    },
]
```

```python
facts = set(['langit_mendung', 'kelembaban_tinggi', 'suhu_rendah'])
print(f"Fakta Awal: {facts}\n")
```

```python
def forward_chaining(initial_facts, knowledge_base):
    current_facts = set(initial_facts)
    new_facts_added = True
    iteration = 0

    while new_facts_added:
        new_facts_added = False
        iteration += 1
        print(f"--- Iterasi {iteration} ---")

        for rule in knowledge_base:
            all_conditions_met = True
            for condition in rule['conditions']:
                if condition not in current_facts:
                    all_conditions_met = False
                    break

            if all_conditions_met and rule['result'] not in current_facts:
                current_facts.add(rule['result'])
                new_facts_added = True
                print(f"Aturan {rule['id']} dieksekusi. Fakta baru ditambahkan: '{rule['result']}' ")

    return current_facts
```

```python
final_facts = forward_chaining(facts, rules)
```

```python
print(f"\nSemua Proses Selesai.")
print(f"Fakta Akhir (Kesimpulan): {final_facts}")
```
